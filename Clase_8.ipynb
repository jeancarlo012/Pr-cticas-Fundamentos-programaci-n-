```python
# 1. Iteraciones con range()
print("--- 1. Iteraciones con range() ---")
print("Cuenta del 0 hasta 7:")
for contador in range(0, 8):
    print(contador)

print("\nCuenta del 5 hasta antes de 9:")
for contador in range(5, 9):
    print(contador)

print("\nCuenta de 3 hasta antes de 11 en incrementos de a 2:")
for contador in range(3, 11, 2):
    print(contador)

print("\nCuenta del 26 hasta más de 10 en decrementos de a -4:")
for contador in range(26, 10, -4):
    print(contador)

# 2. Desempaquetado
print("\n--- 2. Desempaquetado ---")
nombre, apellido, calificacion = ["Juan", "Perez", 7.5]
print(f"Nombre: {nombre}")
print(f"Apellido: {apellido}")
print(f"Calificación: {calificacion}")

# Esta da error ValueError porque son 3 variables para 4 elementos
# nombre, apellido, calificacion = ["Juan", "Perez", "Erroneo", 7.5]

# 3. Desempaquetado con for... in
print("\n--- 3. Desempaquetado con for... in ---")
palabras = ["gato", "pato", "zeta", "cita"]

for item in palabras:
    print(f"Palabra completa: {item}")

print("\nDesempaquetando las letras de cada palabra de 4 letras:")
for primera, segunda, tercera, cuarta in palabras:
    print(primera)
    print(segunda)
    print(tercera)
    print(cuarta)
    print("----------")
```
